# 01 — Download Chinese Traffic Signs Dataset

Downloads the Chinese Traffic Signs dataset from Kaggle using `kagglehub`.

**Note:** Data is saved to `../data/chinese-traffic-signs`.

```python
!pip install -q kagglehub
import os
import csv
import shutil
from pathlib import Path
from PIL import Image
import kagglehub
```

```python
DATA_DIR = Path("../data/chinese-traffic-signs")
DATA_DIR.mkdir(parents=True, exist_ok=True)
print(f"Data directory: {DATA_DIR.resolve()}")
```

```python
print("Downloading Chinese Traffic Signs dataset...")
path = kagglehub.dataset_download("dmitryyemelyanov/chinese-traffic-signs")
print(f"Downloaded to cache: {path}")

for item in os.listdir(path):
    s = os.path.join(path, item)
    d = os.path.join(DATA_DIR, item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)

print("Successfully copied to", DATA_DIR.resolve())
```

```python
annotations = []
with open(DATA_DIR / "annotations.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        annotations.append(row)

print(f"Total images: {len(annotations)}")
num_classes = len(set(row["category"] for row in annotations))
print(f"Num classes: {num_classes}")

sample = annotations[0]
image_path = DATA_DIR / "images" / sample["file_name"]
with Image.open(image_path) as img:
    print(f"Sample image: shape={img.size} (PIL), label={sample['category']}")
```
