# 01 — Download GTSRB Dataset

Downloads the German Traffic Sign Recognition Benchmark (GTSRB) using torchvision.

**Note:** This downloads ~300MB+ of images (plus archive). Data is saved to `../data/`.

```python
from pathlib import Path
import torch
import torchvision
from torchvision.datasets import GTSRB

print(f"torch: {torch.__version__}")
print(f"torchvision: {torchvision.__version__}")
```

```python
DATA_DIR = Path("../data")
DATA_DIR.mkdir(parents=True, exist_ok=True)
print(f"Data directory: {DATA_DIR.resolve()}")
```

```python
train_dataset = GTSRB(root=DATA_DIR, split="train", download=True)
test_dataset = GTSRB(root=DATA_DIR, split="test", download=True)
```

```python
image, label = train_dataset[0]
print(f"Train size: {len(train_dataset)}")
print(f"Test size:  {len(test_dataset)}")
num_classes = len(set(sample[1] for sample in train_dataset._samples))
print(f"Num classes: {num_classes}")
print(f"Sample image: shape={tuple(image.size)} (PIL), label={label}")
```
