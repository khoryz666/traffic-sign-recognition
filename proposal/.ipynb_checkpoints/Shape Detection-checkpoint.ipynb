```python
#Import the libraries
import os     #auto processes every image in subfolders
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
input_folder = "Inputs"

count = 0

for colour in ["Red Signs", "Blue Signs", "Yellow Signs"]:

    folder = os.path.join(input_folder, colour)

    for filename in os.listdir(folder):

        if filename.lower().endswith((".jpg", ".png", ".jpeg")):

            image_path = os.path.join(folder, filename)

            img = cv2.imread(image_path)

            # print(f"Processing {colour}: {filename}")

            # Color conversion
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

            img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

            #Image Preprocessing - Gaussian Blur Filtering
            gaussian_filtered = cv2.GaussianBlur(img_gray, (5,5), 0)
            
    
            # Adaptive Mean Threshold
            adaptive_mean = cv2.adaptiveThreshold(
                gaussian_filtered,
                255,
                cv2.ADAPTIVE_THRESH_MEAN_C,
                cv2.THRESH_BINARY,
                21,   # Must be odd
                5
            )

            # Contour Detection
            contours, hierarchy = cv2.findContours(
                adaptive_mean,
                cv2.RETR_TREE,
                cv2.CHAIN_APPROX_SIMPLE
            )

            #detect tiny contours
            for cnt in contours:

                area = cv2.contourArea(cnt)
            
                if area < 500:
                    continue
            
            #Filter by area
            valid_contours = []

            MIN_AREA = 1000
            
            for cnt in contours:
            
                area = cv2.contourArea(cnt)
            
                if area < MIN_AREA:
                    continue
            
                valid_contours.append(cnt)
                
            #Filter by Circularity
            filtered_contours = []

            for cnt in valid_contours:
            
                area = cv2.contourArea(cnt)
                perimeter = cv2.arcLength(cnt, True)
            
                if perimeter == 0:
                    continue
            
                circularity = 4 * np.pi * area / (perimeter * perimeter)
            
                if circularity > 0.80:
                    filtered_contours.append(cnt)
                    
            #Draw countours
            result = img_rgb.copy()
            best_contour = None
            best_area = 0
            
            for cnt in filtered_contours:
            
                area = cv2.contourArea(cnt)
            
                if area > best_area:
                    best_area = area
                    best_contour = cnt
            
            if best_contour is not None:
                cv2.drawContours(result, [best_contour], -1, (0,255,0), 2)

            # Add an aspect ratio filter 
            x, y, w, h = cv2.boundingRect(cnt)
            
            aspect_ratio = w / float(h)
            
            if not (0.8 <= aspect_ratio <= 1.2):
                continue
            
            
            # Shape Detection
           
                    
            # Display the processing stages
            print(f"Processing {count+1}: {colour} - {filename}")
            
            plt.figure(figsize=(15,4))

            plt.subplot(1,5,1)
            plt.imshow(img_rgb)
            plt.title("Original")
            plt.axis("off")
            
            plt.subplot(1,5,2)
            plt.imshow(img_gray, cmap="gray")
            plt.title("Grayscale")
            plt.axis("off")
            
            plt.subplot(1,5,3)
            plt.imshow(gaussian_filtered, cmap="gray")
            plt.title("Gaussian Blur")
            plt.axis("off")
            
            plt.subplot(1,5,4)
            plt.imshow(adaptive_mean, cmap="gray")
            plt.title("Adaptive Mean")
            plt.axis("off")
            
            plt.subplot(1,5,5)
            plt.imshow(result)
            plt.title("Filtered Contours")
            plt.axis("off")
                        
            plt.suptitle(f"{colour} - {filename}")
            
            plt.show()

            for cnt in filtered_contours:

                area = cv2.contourArea(cnt)
                perimeter = cv2.arcLength(cnt, True)
            
                circularity = 4 * np.pi * area / (perimeter * perimeter)
            
                print(f"Area: {area:.1f}, Circularity: {circularity:.2f}")
    
            #Only show a few images to adjust gaussian 
            count += 1

            if count >= 6:
                break
    # break outer loop
    if count >= 6:
        break


            
```
