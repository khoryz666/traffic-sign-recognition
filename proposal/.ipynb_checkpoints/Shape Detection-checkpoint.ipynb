```python
#Import the libraries
import os     #auto processes every image in subfolders
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
input_folder = "Inputs"

count = 0

for colour in ["Red Signs", "Blue Signs", "Yellow Signs"]:

    folder = os.path.join(input_folder, colour)

    for filename in os.listdir(folder):

        if filename.lower().endswith((".jpg", ".png", ".jpeg")):

            image_path = os.path.join(folder, filename)

            img = cv2.imread(image_path)

            # print(f"Processing {colour}: {filename}")

            # Color conversion
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

            img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

            #Image Preprocessing - Gaussian Blur Filtering
            gaussian_filtered = cv2.GaussianBlur(img_gray, (5,5), 0)
            
    
            # Adaptive Mean Threshold
            adaptive_mean = cv2.adaptiveThreshold(
                gaussian_filtered,
                255,
                cv2.ADAPTIVE_THRESH_MEAN_C,
                cv2.THRESH_BINARY,
                21,   # Must be odd
                5
            )
            
            # Adaptive Gaussian Threshold
            adaptive_gaussian = cv2.adaptiveThreshold(
                gaussian_filtered,
                255,
                cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                cv2.THRESH_BINARY_INV,
                21,
                5
            )

            # Morphological Operations
            kernel = np.ones((5,5), np.uint8)

            adaptive_gaussian = cv2.morphologyEx(
                adaptive_gaussian,
                cv2.MORPH_OPEN,
                kernel
            )
            
            adaptive_gaussian = cv2.morphologyEx(
                adaptive_gaussian,
                cv2.MORPH_CLOSE,
                kernel
            )

            # Contour Detection
            contours, _ = cv2.findContours(
                adaptive_gaussian,
                cv2.RETR_EXTERNAL,
                cv2.CHAIN_APPROX_SIMPLE
            )

            #detect tiny contours
            for cnt in contours:

                area = cv2.contourArea(cnt)
            
                if area < 500:
                    continue
                        # Display the processing stages
            print(f"Processing {count+1}: {colour} - {filename}")
            
            plt.figure(figsize=(18,4))

            plt.subplot(1,6,1)
            plt.imshow(img_rgb)
            plt.title("Original")
            plt.axis("off")
            
            plt.subplot(1,6,2)
            plt.imshow(img_gray, cmap="gray")
            plt.title("Grayscale")
            plt.axis("off")
            
            plt.subplot(1,6,3)
            plt.imshow(gaussian_filtered, cmap="gray")
            plt.title("Gaussian Blur")
            plt.axis("off")
            
            plt.subplot(1,6,4)
            plt.imshow(thresh, cmap="gray")
            plt.title("Otsu")
            plt.axis("off")
            
            plt.subplot(1,6,5)
            plt.imshow(adaptive_mean, cmap="gray")
            plt.title("Adaptive Mean")
            plt.axis("off")
            
            plt.subplot(1,6,6)
            plt.imshow(adaptive_gaussian, cmap="gray")
            plt.title("Adaptive Gaussian")
            plt.axis("off")
            
            plt.suptitle(f"{colour} - {filename}")
            
            plt.show()

            #Only show a few images to adjust gaussian 
            count += 1

            if count >= 6:
                break
    # break outer loop
    if count >= 6:
        break
            #Filter by area

            #Filter by Circularity

            #Draw countours
        
            # Shape Detection


            
```
